```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if문 사용)
total = 0
for i in range (1, 101):
  if i % 2 == 0:
    total = total + i
print (total)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if문 사용 X)
total = 0
for i in range (0, 101, 2):
  total = total + i
print (total)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요.
num = int(input("정수 하나를 입력하세요.: "))
for i in range(1, 10):
  print (f'{num} x { i } = {num*i}')
```

```python
#1부터 100까지의 합을 출력하시오. (while)
a = 1
sum = 0
while a<=100:
  sum = sum + a
  a =a + 1
print(sum)
```

```python
#1부터 100까지의 합을 출력하시오. (while)
a = 1
sum = 0
while True:
  if a<101:
    sum += a
    a += 1
  else:
    break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하시오.
sum = 0
while True:
  num = int(input("양의 정수를 입력하세요: "))
  if num == 0:
    break
  sum += num
print(sum)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 만들어 출력해 보세요.
evens = []
for i in range(1, 21):
  if i % 2 == 0:
    evens.append(i)
print(evens)

evens.pop(7)
print(evens)

del evens[5]
print(evens)
print(evens[3])
print(evens[-1]) #리스트 마지막 항목
```

```python
total=0
max_score=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  total += i
  if i > max_score:
    max_score = i
print(total)
print(max_score)
```
