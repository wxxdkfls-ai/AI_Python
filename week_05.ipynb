```python
#컴퓨터가 1~100사이의 난수 발생, 사용자가 난수를 맞히는 프로그램. 작다/크다/같다 뜨도록
import random
user = int(input('수를 맞혀보세요'))
com = random.randint(1, 100)
print(com)

if user > com:
    print('수가 큽니다.')
elif user < com:
    print('수가 작습니다.')
else:
    print('맞혔습니다.')
```

```python
#asmd 함수 정의
def asmd(x, y, op):
    if op == '+':
        print(f'{x} + {y} = {x+y}')
    elif op == '-':
        print(f'{x} - {y} = {x-y}')
    elif op == '*':
        print(f'{x} * {y} = {x*y}')
    elif op == '/':
        print(f'{x} / {y} = {x/y}')
    else:
        print('잘못된 연산자입니다.')
```

```python
x = int(input('첫 번째 수'))
y = int(input('두 번째 수'))
op =input('연산자를 입력하시오')
asmd(x, y, op)
```

```python
#5번 반복하기
import random
com = random.randint(1, 100)

for i in range(5):
    user = int(input('수를 맞혀보세요'))
    if user > com:
        print('수가 큽니다.')
    elif user < com:
        print('수가 작습니다.')
    else:
        print('맞혔습니다.')
        break

print(com)
```

```python
#while문으로 변경하기
import random
com = random.randint(1, 100)

while True:
    user = int(input('수를 맞혀보세요'))
    if user > com:
        print('수가 큽니다.')
    elif user < com:
        print('수가 작습니다.')
    else:
        print('맞혔습니다.')
        break

print(com)
```

```python
#정답을 맞힐 때까지 얼마나 걸렸는지 출력하기  
import random
com = random.randint(1, 100)
count = 0

while True:
    user = int(input('수를 맞혀보세요'))
    count +=1
    if user > com:
        print('수가 큽니다.')
    elif user < com:
        print('수가 작습니다.')
    else:
        print('맞혔습니다.')
        print(f'시도한 횟수는 {count} 번 입니다.')
        break
        
print(com)
```
