```python
#두 자연수 10과 5가 주어진다.
#이때, A+B, A-B, A*B, A/B(몫), A%B(나머지)를 출력하는 프로그램을 작성하시오.

A = 10
B = 5
print(A+B)
print(A-B)
print(A*B)
print(A/B)
print(A%B)
```

```python
#가로 12, 세로 7인 직사각형의 넓이와 둘레를 구하시오.

width = 12
height = 7
area = width*height
perimeter = width*2+height*2
print(area)
print(perimeter)
```

```python
#국어 88, 영어 92, 수학 75점인 학생의 과목 평균을 구하시오.

Korean = 88
English = 92
Math = 75
average = (Korean+English+Math)/3
print(average)
```
